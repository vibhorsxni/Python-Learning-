```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import time
```

```python
driver = webdriver.Chrome()
```

```python
url = 'https://www.imdb.com/chart/top/'
driver.get(url)
time.sleep(5) 
```

```python
for _ in range(5):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)
```

```python
html = driver.page_source
driver.quit()
```

```python
soup = BeautifulSoup(html,'lxml')
```

```python
rank = []
movie_name = []
release = []
rating = []
review_count = []
licence_type = []
watch_time = []

movies = soup.find_all(class_="ipc-metadata-list-summary-item")
```

```python
for movie in movies :
    # Ranks
    try :
        rank.append(movie.find('div',class_="ipc-signpost__text").text)
    except :
        rank.append(np.nan)

    # Shows_Name
    try :
        movie_name.append(movie.find('h4',class_="ipc-title__text").text.strip())
    except:
        movie_name.append(np.nan)

    # Release Year 
    try :
        release.append(movie.find('li',class_="ipc-inline-list__item").text.strip())
    except:
        release.append(np.nan)

    # Rating 
    try :
        rating.append(movie.find('span' ,class_='ipc-rating-star--rating').text.strip())
    except:
        rating.append(np.nan)

    # Review_Count
    try :
        review_count.append(movie.find('span',class_="ipc-rating-star--voteCount").text.strip())
    except:
        review_count.append(np.nan)

    meta_items = movie.find_all('li', class_="ipc-inline-list__item")

    # licence_type
    try :
        licence_type.append(meta_items[2].text.strip())
    except:
        licence_type.append(np.nan)
        
    # watch_time
    try :
        watch_time.append(meta_items[1].text.strip())
    except:
        watch_time.append(np.nan)
    
```

```python
print(len(rank))
print(len(movie_name))
print(len(release))
print(len(rating))
print(len(review_count))
print(len(licence_type))
print(len(watch_time))
```

```python
df = pd.DataFrame({'Ranking':rank,
'Movie':movie_name,
'Release Year' :release,
'Star Rating' :rating,
'Review Count' : review_count,
'Licence Category' : licence_type,
'Movie Length ' : watch_time })
```
