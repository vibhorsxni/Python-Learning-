### Groupby() Function -

```python
delivery = pd.read_csv('deliveries.csv')
```

```python
delivery.groupby('batsman')['batsman_runs'].count().sort_values(ascending = False)
```

```python
 delivery.head(15)
```

```python
vk = delivery[delivery['batsman'] == 'V Kohli']
```

```python
vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending = False)
```

```python
def max_run_opp_team(batsman) :
  x = delivery[delivery['batsman'] == batsman]
  return x.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending = False).index[0]
```

```python
max_run_opp_team('DA Warner')
```
