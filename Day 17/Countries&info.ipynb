```python
all_countries = []
for offset in range(0, 300, 100):
    api_url = f"https://api.restcountries.com/countries/v5?limit=100&offset={offset}"
    response = requests.get(api_url, headers=api_headers)
    raw_json = response.json()
    if 'data' not in raw_json or len(raw_json['data']['objects']) == 0:
        break
        
    all_countries.extend(raw_json['data']['objects'])
    
```

```python
df = pd.DataFrame(all_countries)
df.shape
```

```python
df['Country'] = df['names'].apply(lambda x: x['common'])
df['Capital'] = df['capitals'].apply(lambda x: x[0]['name'] if len(x) > 0 else None)
df['Capital-Lat'] = df['capitals'].apply(lambda x: x[0]['coordinates']['lat'] if len(x) > 0 else None)
df['Capital-Lng'] = df['capitals'].apply(lambda x: x[0]['coordinates']['lng'] if len(x) > 0 else None)
df['Population'] = df['population']
df['Region'] = df['region']
df['Sub-Region'] = df['subregion']
df['Timezone'] = df['timezones']
df['Area(km)'] = df['area'].apply(lambda x: x['kilometers'])
df['Border Nation'] = df['borders']
df['Government'] = df['government_type']
df['Call - Codes'] = df['calling_codes'].apply(lambda x: x[0] if x else None)
df['Continent'] = df['continents'].apply(lambda x: x[0] if x else None)
df['Currency_Code'] = df['currencies'].apply(lambda x: x[0]['code'] if x else None)
df['Currency_Name'] = df['currencies'].apply(lambda x: x[0]['name'] if x else None)
df['Currency_Symbol'] = df['currencies'].apply(lambda x: x[0]['symbol'] if x else None)
df['Languages'] = df['languages'].apply(lambda x: ', '.join(lang['name'] for lang in x) if x else None)
df['Driving_Side'] = df['cars'].apply(lambda x: x['driving_side'] if x else None)


clean_df = df[[
    'Country',
    'Capital',
    'Capital-Lat',
    'Capital-Lng',
    'Population',
    'Area(km)',
    'Continent',
    'Region',
    'Sub-Region',
    'Languages',
    'Driving_Side',
    'Government',
    'Currency_Code',
    'Currency_Name',
    'Currency_Symbol',
    'Call - Codes',
    'Timezone',
    'Border Nation'
]]

clean_df = clean_df.set_index('Country')
clean_df
```

```python
clean_df.to_csv('countries-info.csv')
```
