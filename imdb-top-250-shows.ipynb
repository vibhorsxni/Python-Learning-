```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import time
```

```python
driver = webdriver.Chrome()
```

```python
url = 'https://www.imdb.com/chart/toptv/'
driver.get(url)
time.sleep(5) 
```

```python
for _ in range(5):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(2)
```

```python
html = driver.page_source
driver.quit()
```

```python
soup = BeautifulSoup(html,'lxml')
```

```python
rank = []
show_name = []
release = []
rating = []
review_count = []
licence_type = []
series_type = []

shows = soup.find_all(class_="ipc-metadata-list-summary-item")
```

```python
for show in shows :
    # Ranks
    try :
        rank.append(show.find('div',class_="ipc-signpost__text").text)
    except :
        rank.append(np.nan)

    # Shows_Name
    try :
        show_name.append(show.find('h4',class_="ipc-title__text").text.strip())
    except:
        show_name.append(np.nan)

    # Release Year 
    try :
        release.append(show.find('li',class_="ipc-inline-list__item").text.strip())
    except:
        release.append(np.nan)

    # Rating 
    try :
        rating.append(show.find('span' ,class_='ipc-rating-star--rating').text.strip())
    except:
        rating.append(np.nan)

    # Review_Count
    try :
        review_count.append(show.find('span',class_="ipc-rating-star--voteCount").text.strip())
    except:
        review_count.append(np.nan)

    meta_items = show.find_all('li', class_="ipc-inline-list__item")

    # licence_type
    try :
        licence_type.append(meta_items[1].text.strip())
    except:
        licence_type.append(np.nan)
        
    # Series_type
    try :
        series_type.append(meta_items[2].text.strip())
    except:
        series_type.append(np.nan)
    
```

```python
print(len(rank))
print(len(show_name))
print(len(release))
print(len(rating))
print(len(review_count))
print(len(licence_type))
print(len(series_type))
```

```python
df = pd.DataFrame({'Ranking':rank,
'Movie':show_name,
'Release Year' :release,
'Star Rating' :rating,
'Review Count' : review_count,
'Licence Category' : licence_type,
'Show Type ' : series_type })
```
